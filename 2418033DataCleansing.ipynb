## **4. Data Cleansing-Standarisasi kolom ID_Pemain**

```python
# Function to format 'ID Pemain'
def format_ID_Pemain(ID):
    if ID.startswith("PM-") and len(ID.split('-')[-1]) == 4:
        return ID[:3] + '0' + ID[3:]  # Add an extra '0' after "ID-"
    return ID
```

Program tersebut digunakan untuk menyeragamkan format data pada kolom ID_Pemain agar memiliki format ID yang konsisten. Fungsi format_ID_Pemain(ID) digunakan untuk memeriksa setiap ID pemain. Perintah startswith("PM-") digunakan untuk memastikan ID diawali dengan PM-, sedangkan len(ID.split('-')[-1]) == 4 digunakan untuk memeriksa apakah angka setelah tanda - terdiri dari 4 digit. Jika kedua kondisi terpenuhi, ID[:3] + 0 + ID[3:] digunakan untuk menambahkan angka 0 setelah PM-, misalnya PM-1234 menjadi PM-01234. Jika kondisi tidak terpenuhi, return ID mengembalikan ID tanpa perubahan. Selanjutnya, apply(format_ID_Pemain) menerapkan fungsi tersebut ke seluruh data pada kolom ID_Pemain.


```python
# Apply the function to the 'ID_Pemain' column
data['ID_Pemain'] = data['ID_Pemain'].apply(format_ID_Pemain)
```

Perintah tersebut digunakan untuk menerapkan fungsi format_ID_Pemain pada seluruh data yang terdapat di kolom ID_Pemain. Bagian data['ID_Pemain'] digunakan untuk memilih kolom ID_Pemain, kemudian apply(format_ID_Pemain) menjalankan fungsi format_ID_Pemain pada setiap nilai ID. Hasil dari proses tersebut kemudian disimpan kembali ke kolom ID_Pemain, sehingga setiap ID pemain yang belum sesuai format akan diperbaiki sesuai aturan yang telah dibuat.


```python
# Display the updated 'ID_Pemain' column
data[['ID_Pemain']].head()
```

### **5. Data Cleansing-Standarisasi kolom Nama_Pemain**

```python
# Function to clean 'Nama_Pemain'
def clean_name(name):
    name = str(name).strip() #Menghilangkan spasi di awal dan akhir.
    name = ' '.join(name.split()) #Menghilangkan spasi berlebihan.
    return name.title()

# Apply the cleaning function
data['Nama_Pemain'] = data['Nama_Pemain'].apply(clean_name)

# Display the cleaned data
data[['Nama_Pemain']].head()
```

Program tersebut digunakan untuk membersihkan dan menyeragamkan data pada kolom Nama_Pemain. Fungsi clean_name(name) digunakan untuk membersihkan setiap nama pemain. Perintah str(name) mengubah data menjadi teks, sedangkan strip() digunakan untuk menghilangkan spasi di awal dan akhir nama. Perintah split() dan join() digunakan untuk menghilangkan spasi berlebihan di antara kata. Kemudian title() digunakan untuk membuat huruf awal setiap kata menjadi huruf kapital. Selanjutnya, apply(clean_name) menerapkan fungsi tersebut ke seluruh data pada kolom Nama_Pemain. Terakhir, data[['Nama_Pemain']].head() digunakan untuk menampilkan 5 data pertama setelah proses pembersihan.


## **6. Data Cleansing-Standarisasi Kolom Usia**

```python
# Function to standardize 'Usia'
def standardize_age(age):
    age = str(age)
    age = age.replace('tahun', '')
    age = age.replace('thn', '')
    age = age.replace('th', '')
    age = age.replace(' ', '')
    return int(age)

# Apply the standardization function
data['Usia'] = data['Usia'].apply(standardize_age)

# Display the standardized age
data[['Usia']].head()
```

Program tersebut digunakan untuk menyeragamkan format data pada kolom Usia agar seluruh data usia menjadi angka. Fungsi standardize_age(age) digunakan untuk membersihkan data usia dengan mengubahnya menjadi teks menggunakan str(age). Perintah replace('tahun', ''), replace('thn', ''), dan replace('th', '') digunakan untuk menghapus berbagai bentuk satuan usia. Selanjutnya, replace(' ', '') digunakan untuk menghilangkan spasi pada data. Perintah int(age) mengubah hasilnya menjadi bilangan bulat, misalnya 21 tahun menjadi 21. Kemudian apply(standardize_age) menerapkan fungsi tersebut ke seluruh data pada kolom Usia. Terakhir, data[['Usia']].head() digunakan untuk menampilkan 5 data pertama setelah proses standardisasi.

## **10. Data Cleansing Standarisasi kolom Penampilan**

```python
# Function to standardize 'Penampilan'
def standardize_penampilan(appearance):
    appearance = str(appearance)
    appearance = appearance.replace('x/musim', '')
    appearance = appearance.replace('kali', '')
    appearance = appearance.replace('x', '')
    appearance = appearance.strip()
    return int(appearance)

# Apply the standardization function
data['Penampilan'] = data['Penampilan'].apply(standardize_penampilan)

# Display the standardized data
data[['Penampilan']].head()
```

Program tersebut digunakan untuk menyeragamkan format data pada kolom Penampilan agar seluruh data penampilan menjadi angka. Fungsi standardize_penampilan(appearance) digunakan untuk membersihkan data penampilan dengan mengubahnya menjadi teks menggunakan str(appearance). Perintah replace('x/musim', ''), replace('kali', ''), dan replace('x', '') digunakan untuk menghapus satuan atau keterangan yang terdapat pada data, seperti 23x, 27 kali, atau 23x/musim. Selanjutnya, strip() digunakan untuk menghilangkan spasi di awal dan akhir data. Perintah int(appearance) mengubah hasilnya menjadi bilangan bulat. Kemudian apply(standardize_penampilan) menerapkan fungsi tersebut ke seluruh data pada kolom Penampilan. Terakhir, data[['Penampilan']].head() digunakan untuk menampilkan 5 data pertama setelah proses standardisasi.

# **13. Data Cleansing Standarisasi kolom Menit_Bermain**

Perintah int(minute) mengubah hasilnya menjadi bilangan bulat. Kemudian apply(standardize_menit) menerapkan fungsi tersebut ke seluruh data pada kolom Menit_Bermain.
