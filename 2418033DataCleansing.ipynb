## **4. Data Cleansing-Standarisasi kolom ID_Pemain**

```python
# Function to format 'ID Pemain'
def format_ID_Pemain(ID):
    if ID.startswith("PM-") and len(ID.split('-')[-1]) == 4:
        return ID[:3] + '0' + ID[3:]  # Add an extra '0' after "ID-"
    return ID
```

```python
# Apply the function to the 'ID_Pemain' column
data['ID_Pemain'] = data['ID_Pemain'].apply(format_ID_Pemain)
```

```python
# Display the updated 'ID_Pemain' column
data[['ID_Pemain']].head()
```

### **5. Data Cleansing-Standarisasi kolom Nama_Pemain**

```python
# Function to clean 'Nama_Pemain'
def clean_name(name):
    name = str(name).strip() #Menghilangkan spasi di awal dan akhir.
    name = ' '.join(name.split()) #Menghilangkan spasi berlebihan.
    return name.title()

# Apply the cleaning function
data['Nama_Pemain'] = data['Nama_Pemain'].apply(clean_name)

# Display the cleaned data
data[['Nama_Pemain']].head()
```

## **6. Data Cleansing-Standarisasi Kolom Usia**

```python
# Function to standardize 'Usia'
def standardize_age(age):
    age = str(age)
    age = age.replace('tahun', '')
    age = age.replace('thn', '')
    age = age.replace('th', '')
    age = age.replace(' ', '')
    return int(age)

# Apply the standardization function
data['Usia'] = data['Usia'].apply(standardize_age)

# Display the standardized age
data[['Usia']].head()
```

## **7. Data Cleansing Standarisasi kolom Jenis_Kelamin**

```python
# Standardize 'Jenis_Kelamin'
data['Jenis_Kelamin'] = data['Jenis_Kelamin'].replace({
    'Laki laki': 'Laki-laki'
})

# Display the standardized data
data[['Jenis_Kelamin']].head()
```

## **8. Data Cleansing Standarisasi kolom Posisi**

```python
# Standardize 'Posisi'
data['Posisi'] = data['Posisi'].str.title()

# Display the standardized data
data[['Posisi']].head()
```

# **9. Data Cleansing Standarisasi kolom Kaki_Dominan**

```python
# Standardize 'Kaki_Dominan'
data['Kaki_Dominan'] = data['Kaki_Dominan'].str.title()

# Display the standardized data
data[['Kaki_Dominan']].head()
```

## **10. Data Cleansing Standarisasi kolom Penampilan**

```python
# Function to standardize 'Penampilan'
def standardize_penampilan(appearance):
    appearance = str(appearance)
    appearance = appearance.replace('x/musim', '')
    appearance = appearance.replace('kali', '')
    appearance = appearance.replace('x', '')
    appearance = appearance.strip()
    return int(appearance)

# Apply the standardization function
data['Penampilan'] = data['Penampilan'].apply(standardize_penampilan)

# Display the standardized data
data[['Penampilan']].head()
```

# **11. Data Cleansing Standarisasi kolom Gol**

```python
# Replace '-' with 0 in 'Gol'
data['Gol'] = data['Gol'].replace('-', 0)

# Convert to integer
data['Gol'] = data['Gol'].astype(int)

# Display the cleaned data
data[['Gol']].head()
```

# **12. Data Cleansing Standarisasi kolom Assist**

```python
# Replace '-' with 0 in 'Assist'
data['Assist'] = data['Assist'].replace('-', 0)

# Convert to integer
data['Assist'] = data['Assist'].astype(int)

# Display the cleaned data
data[['Assist']].head()
```

# **13. Data Cleansing Standarisasi kolom Menit_Bermain**

```python
# Function to standardize 'Menit_Bermain'
def standardize_menit(minute):
    minute = str(minute)
    minute = minute.replace('menit', '')
    minute = minute.replace('mnt', '')
    minute = minute.strip()
    return int(minute)

# Apply the standardization function
data['Menit_Bermain'] = data['Menit_Bermain'].apply(standardize_menit)

# Display the standardized data
data[['Menit_Bermain']].head()
```

# **14. Data Cleansing Standarisasi kolom Status**

```python
# Standardize 'Status'
data['Status'] = data['Status'].str.title()

# Display the standardized data
data[['Status']].head()
```

# **15. Data Cleansing-Deduplikasi Data**

```python
# Check duplicate data
duplicate_data = data[data.duplicated(keep=False)]

print("Data Duplikat:")
print(duplicate_data)
```
